```python
# Question 1: Generate a list of 100 integers containing values between 90 to 130 and store it in the variable int_list. Find the mean, median, mode, weighted mean, geometric mean, harmonic mean, midrange, and trimmed mean.

import numpy as np
from scipy import stats

np.random.seed(42)
int_list = np.random.randint(90, 131, 100)

mean_value = np.mean(int_list)
median_value = np.median(int_list)
mode_value = stats.mode(int_list, keepdims=True).mode[0]

weights = np.ones(len(int_list))
weighted_mean = np.average(int_list, weights=weights)

geometric_mean = stats.gmean(int_list)
harmonic_mean = stats.hmean(int_list)
midrange = (np.min(int_list) + np.max(int_list)) / 2
trimmed_mean = stats.trim_mean(int_list, 0.10)

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
print("Weighted Mean:", weighted_mean)
print("Geometric Mean:", geometric_mean)
print("Harmonic Mean:", harmonic_mean)
print("Midrange:", midrange)
print("Trimmed Mean:", trimmed_mean)
```

```python
# Question 10: Write a Python class representing a discrete random variable with methods to calculate its expected value and variance.

class DiscreteRandomVariable:
    def __init__(self, values, probabilities):
        self.values = np.array(values)
        self.probabilities = np.array(probabilities)

    def expected_value(self):
        return np.sum(self.values * self.probabilities)

    def variance(self):
        mean = self.expected_value()
        return np.sum((self.values - mean) ** 2 * self.probabilities)

values = [1, 2, 3, 4, 5, 6]
probabilities = [1/6] * 6

rv = DiscreteRandomVariable(values, probabilities)

print("Expected Value:", rv.expected_value())
print("Variance:", rv.variance())
```

```python
# Question 12: Create a Python function to generate random samples from a given probability distribution and calculate their mean and variance.

def generate_samples(distribution, size=1000):
    if distribution == "binomial":
        samples = np.random.binomial(n=10, p=0.5, size=size)
    elif distribution == "poisson":
        samples = np.random.poisson(lam=5, size=size)
    else:
        raise ValueError("Use 'binomial' or 'poisson'.")

    return samples, np.mean(samples), np.var(samples)

samples, sample_mean, sample_variance = generate_samples("poisson")

print("Mean:", sample_mean)
print("Variance:", sample_variance)
```

```python
# Question 17: Use seaborn library to load the tips dataset. Write a Python function that calculates skewness for total_bill and tip.

tips = sns.load_dataset("tips")

def calculate_skewness(data):
    return data.skew()

print("Total Bill Skewness:", calculate_skewness(tips["total_bill"]))
print("Tip Skewness:", calculate_skewness(tips["tip"]))
```

```python
# Question 18: Determine whether the total_bill and tip columns exhibit positive skewness, negative skewness, or are approximately symmetric.

def skewness_type(value):
    if value > 0.1:
        return "Positive skewness"
    elif value < -0.1:
        return "Negative skewness"
    return "Approximately symmetric"

print("Total Bill:", skewness_type(tips["total_bill"].skew()))
print("Tip:", skewness_type(tips["tip"].skew()))
```

```python
# Question 19: Write a function that calculates the covariance between two columns.

def calculate_covariance(x, y):
    return np.cov(x, y)[0, 1]

print("Covariance:", calculate_covariance(tips["total_bill"], tips["tip"]))
```

```python
# Question 23: Perform a Chi-Square test to determine if there is a significant difference between job performance levels before and after the training.

from scipy.stats import chi2_contingency

data = np.array([
    [50, 30, 20],
    [30, 40, 30],
    [20, 30, 40]
])

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)

if p_value < 0.05:
    print("There is a significant difference.")
else:
    print("There is no significant difference.")
```

```python
# Question 24: Perform an ANOVA test to determine if there is a significant difference in customer satisfaction scores among the three product versions.

standard_scores = [80, 85, 90, 78, 88, 82, 92, 78, 85, 87]
premium_scores = [90, 92, 88, 92, 95, 91, 96, 93, 89, 93]
deluxe_scores = [95, 98, 92, 97, 96, 94, 98, 97, 92, 99]

f_statistic, p_value = stats.f_oneway(
    standard_scores,
    premium_scores,
    deluxe_scores
)

print("F-statistic:", f_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("There is a significant difference between the groups.")
else:
    print("There is no significant difference between the groups.")
```
